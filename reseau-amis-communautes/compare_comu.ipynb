2) Comparer théoriquement et pratiquement les fonctions cree_reseau et dico_reseau. (Cette dernière construit le réseau en cherchant d'abord les amis d'Alice, puis de Bob, etc.)

Comparaison théorique : En comparant ces deux fonctions nous remarquons d'importantes différences plus précisément sur la notion de complexité. La fonction cree_reseau adopte une stratégie directe : elle parcourt une seule fois la liste des couples d'amis, et pour chaque couple (a,b), elle effectue seulement des opérations élémentaires en temps constant - à savoir deux insertions dans les listes d'amis et des tests d'appartenance sur le dictionnaire. Cette approche conduit à une complexité linéaire O(n) où n est le nombre de couples d'amis. Tandis que dico_reseau fonctionne en deux temps : elle doit d'abord extraire la liste complète des personnes, ce qui nécessite un premier parcours, puis pour chaque personne identifiée, elle doit reparcourir toute la liste des couples pour trouver tous ses amis. Cela conduit alors à une complexité quadratique O(n*p), où n est le nombre de couples et p le nombre de personnes. Cette différence de complexité théorique s'explique par le fait que dico_reseau effectue des recherches répétitives/redondantes là où cree_reseau construit directement les relations au fil de la lecture des données.
Comparaison pratique : Pour comparer les fonctions en pratique, nous avons créé un petit réseau de test avec 10 personnes (A à J) reliées par 16 couples d'amis.


```python
import time                                             # Importation du module pour mesurer le temps d'exécution
from comu import cree_reseau



def personnes(amis):
    """
    Retourne un tableau contenant les différentes personnes présentes dans le tableau `amis`.
    
    Paramètre :
    amis : tableau des relations d'amitié où chaque élément est une chaîne de caractères représentant une personne.
    
    Retourne :
    Un tableau contenant les prénoms uniques des personnes dans `amis`.
    """
    personnes = []                                      # Liste pour stocker les noms uniques des personnes
    i = 0                                               # Initialisation de l'indice pour parcourir `amis`
    while i < len(amis):                                # Boucle sur tous les éléments du tableau `amis`
        if amis[i] not in personnes:                    # Si la personne n'est pas déjà dans la liste
            personnes.append(amis[i])                   # On l'ajoute à la liste des personnes
        i += 1                                          # On passe à l'élément suivant
    return personnes                                    # Retourne la liste des personnes


def ses_amis(amis, prenom):
    """
    Retourne un tableau contenant les amis d'une personne donnée.
    
    Paramètres :
    amis : tableau des relations d'amitié (tableau plat où chaque couple d'indices représente une relation).
    prenom : prénom de la personne pour laquelle on cherche les amis.
    
    Retourne :
    Un tableau contenant les prénoms des amis de `prenom`.
    """
    ses_amis = []                                       # Liste pour stocker les amis de `prenom`
    i = 0                                               # Initialisation de l'indice pour parcourir les relations d'amitié
    while i < len(amis) / 2:                            # Parcourt les relations deux par deux (chaque paire représente une amitié)
        if amis[2 * i] == prenom:                       # Si la première personne de la relation est `prenom`
            ses_amis.append(amis[2 * i + 1])            # On ajoute la deuxième personne comme ami
        elif amis[2 * i + 1] == prenom:                 # Si la deuxième personne de la relation est `prenom`
            ses_amis.append(amis[2 * i])                # On ajoute la première personne comme ami
        i += 1                                          # Passe à la relation suivante
    return ses_amis                                     # Retourne la liste des amis de `prenom`


def dico_reseau(amis):
    """
    Construit un dictionnaire représentant un réseau d'amis.
    
    Paramètre :
    amis : tableau des relations d'amitié (tableau plat où chaque couple d'indices représente une relation).
    
    Retourne :
    Un dictionnaire où chaque clé est un prénom et la valeur associée est un tableau des amis de cette personne.
    """
    reseau = {}                                         # Initialise le dictionnaire du réseau
    pers = personnes(amis)                              # Liste des membres uniques du réseau obtenue avec la fonction `personnes`
    i = 0                                               # Initialisation de l'indice pour parcourir les membres du réseau
    while i < len(pers):                                # Boucle sur chaque personne dans le réseau
        reseau[pers[i]] = ses_amis(amis, pers[i])       # Ajoute les amis de la personne actuelle au dictionnaire
        i += 1                                          # Passe à la personne suivante
    return reseau                                       # Retourne le dictionnaire du réseau

                                          

                                                        # Création d'un tableau test représentant un réseau de 10 personnes
test_couples = [
    ("A", "B"), ("A", "C"), ("B", "C"), ("B", "D"),
    ("C", "D"), ("D", "E"), ("E", "F"), ("F", "G"),
    ("G", "H"), ("H", "I"), ("I", "J"), ("A", "D"),
    ("B", "E"), ("C", "F"), ("D", "G"), ("E", "H")
] * 1000                                                 # On multiplie les relations par 1000 pour créer un réseau plus grand

                                                         # Test de la fonction cree_reseau 
debut = time.time()                                      # Enregistre le temps de départ
res1 = cree_reseau(test_couples)                         # Exécute la fonction cree_reseau
temps1 = time.time() - debut                             # Calcule le temps écoulé
print("cree_reseau met", round(1000 * temps1, 2), "millisecondes")  # Affiche le temps en millisecondes

                                                         # Test de la fonction dico_reseau 
debut = time.time()                                      # Enregistre le temps de départ
res2 = dico_reseau(test_couples)                         # Exécute la fonction dico_reseau
temps2 = time.time() - debut                             # Calcule le temps écoulé
print("dico_reseau met", round(1000 * temps2, 2), "millisecondes")  # Affiche le temps en millisecondes
```

11) Comparer théoriquement et pratiquement les fonctions comu_dans_reseau et comu_dans_amis appliquée à une personne des plus populaires (la recherche de la personne la plus populaire sera prise en compte dans la complexité).

**reponse:** théoriquement: la complexité de comu_dans_reseau est en deux parties, d'abord, on trie le groupe par popularité, donc on a O(nlogn) où n correspond au nombre de personnes dans le groupe. Ensuite on a la construction de la communauté, donc on a O(n** 2). Au total, la complexité de comu_dans_reseau est O(nlogn + n ** 2)

Pour comu_dans_amis, on doit d'abord trouver la personne la plus populaire: O(n), ensuite on trie ses amis par popularité donc O(nlogn) puis on construit la communauté donc O((n + 1)** 2), au total, on a O(n + nlogn + (n + 1)** 2).

pratique : Si on compare les deux fonctions, comu_dans_reseau mets plus de temps si la communauté est grande, et comu_dans_amis est très efficace si la personne a peu d'amis.

```python
import time
from comu import cree_reseau, tri_popu, comu

new_tab= cree_reseau(amis)
def comu_dans_reseau(reseau):
    groupe = list(reseau.keys())
    groupe_trie = tri_popu(reseau, groupe)
    return comu(groupe_trie, reseau)



debut = time.time()
comu_dans_reseau(new_tab)
temps3 = time.time() - debut


def comu_dans_amis(personne, reseau):
    # Obtenir les amis de la personne de départ
    amis = reseau[personne]
    
    # Trier les amis par popularité décroissante en utilisant tri_popu
    amis_tries = tri_popu(reseau, amis)
    
    # Ajouter la personne de départ au début de la liste des amis triés
    groupe = [personne] + amis_tries
    
    # Utiliser la fonction comm pour construire la communauté
    return comu(groupe, reseau)



debut = time.time()
comu_dans_amis("Alice", new_tab)
temps4 = time.time() - debut

print("comu_dans_reseau mets", temps3, "secondes")
print("comu_dans_amis mets", temps4, "secondes")
```
